```python
"""
this section is what I used when building this, so that way the inputs above did not need to be reran every time, 
alongside a confirmation of values below
total_repaid = 0 
interest = 0.05
principal = 500000
year = 30
period = 12
periodic_interest = interest/period
remaining_balance = principal
interest_payment = (remaining_balance * periodic_interest)

total_payment = principal*((interest/period)*(1+interest/period)**(period*year))/((1+interest/period)**(year*period)-1)
"""
print(f"interest: {interest},\nprincipal: {principal},\nyear: {year},\nperiod: {period},\nperiodic interest: {periodic_interest},\ntotal payment: {total_payment},\ninterest payment for this first payment: {interest_payment}")
     


```

```python

records = []

for n in range(year):
    print(n)
    for p in range(period):
        interest_payment = (remaining_balance * periodic_interest)
        principal_payment = (total_payment - interest_payment)
        total_repaid += principal_payment
        remaining_balance -= principal_payment
        period_pay = interest_payment+principal_payment
        print(remaining_balance, principal_payment,interest_payment, total_repaid, period_pay)
#creating dataframe
        records.append({
            "Year": n,
            "Period": p,
            "Payment": round(period_pay, 2),
            "Principal Paid": round(principal_payment, 2),
            "Interest Paid": round(interest_payment, 2),
            "Total Repaid": round(total_repaid, 2),
            "Remaining Balance": round(remaining_balance, 2)
 })



```
